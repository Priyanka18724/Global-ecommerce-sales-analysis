### Calculate Average Order Value

```python
df['Total_Sales'].mean()
```

### Find the hightest-selling Product 

```python
df.groupby("Product_Name")["Total_Sales"].sum().sort_values(ascending=False).head(10)
```

### Find the Profitable Product

```python
df.groupby("Product_Name")["Profit"].sum().sort_values(ascending=False).head(10)
```

### Sales by Product Category

```python
df.groupby("Product_Category")["Total_Sales"].sum().sort_values(ascending=False)
```

### Profit by Product Category

### Sales by Region

```python
df.groupby("Region")["Total_Sales"].sum().sort_values(ascending=False)
```

### Profit by Region

```python
df.groupby("Region")["Profit"].sum().sort_values(ascending=False)
```

### Sales by Customer Segment

### Profit by Customer segment

*** count of Payment Method ***

```python
df["Payment_Method"].value_counts()
```

*** Monthly Sales Analysis ***

```python
monthly_sales = df.groupby(df["Order_Date"].dt.to_period("M"))["Total_Sales"].sum()
monthly_sales
```

*** Find the highest-sales month ***

```python
monthly_sales.idxmax(), monthly_sales.max()
```

*** Discount vs Profit ***

```python
df.groupby("Discount_Percent")["Profit"].sum().sort_index()
```

*** find the highest-profit discount level ***

```python
df.groupby("Discount_Percent")["Profit"].sum()
```

```python
df.groupby("Discount_Percent")["Profit"].sum().idxmax()
```

```python
*** Check shipping cost by region ***
```

```python
df.groupby("Region")["Shipping_Cost"].sum().sort_values(ascending=False)
```

*** Check the most profitable order ***

```python
df.loc[df["Profit"].idxmax()]
```

*** Check the most expensive order ***

```python
df.loc[df["Unit_Price"].idxmax()]
```

*** Create a Sales Summary Tables ***

```python
sales_summary = {
    "Total Orders": len(df),
    "Total Quantity Sold": df["Quantity"].sum(),
    "Total Sales": df["Total_Sales"].sum(),
    "Total Profit": df["Profit"].sum(),
    "Average Order Value": df["Total_Sales"].mean()
}
sales_summary
```

### Visualization

```python
category_sales = df.groupby("Product_Category")["Total_Sales"].sum()
category_sales.plot(kind="bar", title="Sales by Product Category")
```

```python
category_profit = df.groupby("Product_Category")["Profit"].sum()
category_profit.plot(kind="bar", title="Profit by Product Category")
```

```python
monthly_sales.plot(kind="line", title="Monthly Sales Analysis")
```

```python
region_summary = df.groupby("Region")[["Total_Sales", "Profit"]].sum()
region_summary.plot(kind="bar", title="Sales & Profit by Region")
```

```python
discount_profit = df.groupby("Discount_Percent")["Profit"].sum()
discount_profit.plot(kind="line", marker="o", title="Discount vs Profit")
```

```python
shipping_region = df.groupby("Region")["Shipping_Cost"].sum()
shipping_region.plot(kind="bar", title="Shipping Cost by Region")
```

```python
segment_sales = df.groupby("Customer_Segment")["Total_Sales"].sum()
segment_sales.plot(kind="bar", title="Sales by Customer Segment")
```

```python
payment_counts = df["Payment_Method"].value_counts()
payment_counts.plot(kind="bar", title="Orders by Payment Method")
```
